```python
import heapq

def manhattan(state, goal):
    distance = 0

    for i in range(9):
        tile = state[i]

        if tile == '0':
            continue

        goal_pos = goal.index(tile)

        distance += abs(i // 3 - goal_pos // 3)
        distance += abs(i % 3 - goal_pos % 3)

    return distance

def neighbors(state):
    result = []
    zero = state.index('0')
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_zero = nr * 3 + nc
            temp = list(state)
            temp[zero], temp[new_zero] = temp[new_zero], temp[zero]
            result.append(''.join(temp))

    return result

def display(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])

def astar(start, goal):
    pq = []
    h = manhattan(start, goal)

    heapq.heappush(pq, (h, 0, start))
    visited = set()
    iteration = 0

    while pq:
        f, g, state = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)
        h = manhattan(state, goal)
        iteration += 1

        print("\nIteration:", iteration)
        display(state)
        print("G(n) =", g)
        print("H(n) =", h)
        print("F(n) = G(n) + H(n) =", f)

        if state == goal:
            print("\nGoal Reached")
            print("Total Moves:", g)
            return

        for next_state in neighbors(state):
            if next_state not in visited:
                new_g = g + 1
                new_h = manhattan(next_state, goal)
                new_f = new_g + new_h

                heapq.heappush(pq, (new_f, new_g, next_state))

    print("\nNo solution exists")

print("8-Puzzle using A* Algorithm")
print("Enter 0 for the blank space")

start = input("Enter initial state: ")
goal = input("Enter goal state: ")

astar(start, goal)

print("\nName: Anish Balabattuni")
print("USN: 1WN24CS039")
```
